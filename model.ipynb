```python
def spatial_join_companies_lsoa(df_companies, gdf_lsoa, geometry_col='geom'):
    # 创建GeoDataFrame
    gdf_companies = gpd.GeoDataFrame(df_companies, geometry=geometry_col)
    gdf_companies.set_crs('EPSG:4326', inplace=True, allow_override=True)
    
    # 统一CRS
    if gdf_companies.crs != gdf_lsoa.crs:
        gdf_companies = gdf_companies.to_crs(gdf_lsoa.crs)
    
    # 进行空间连接 - 将点数据与包含它们的多边形连接
    merged_data = gpd.sjoin(gdf_companies, gdf_lsoa, how='left', predicate='within')
    
    return merged_data
```

```python
def add_timepoint_field(gdf):
    """
    根据series和epoch列的组合添加timepoint字段
    """
    # 创建条件映射
    conditions = [
        (gdf['series'] == 2010) & (gdf['epoch'] == 21),
        (gdf['series'] == 2010) & (gdf['epoch'] == 23),
        (gdf['series'] == 2017) & (gdf['epoch'] == 2),
        (gdf['series'] == 2017) & (gdf['epoch'] == 4),
        (gdf['series'] == 2017) & (gdf['epoch'] == 7),
        (gdf['series'] == 2017) & (gdf['epoch'] == 13)
    ]
    
    # 对应的时间点标签
    timepoint_labels = [
        '-12',
        '-6', 
        '0',
        '+6',
        '+12',
        '+24'
    ]
    
    # 使用numpy.select创建新字段
    import numpy as np
    gdf['timepoint'] = np.select(conditions, timepoint_labels, default='other')
    
    return gdf

gdf_with_timepoint = add_timepoint_field(business17)
print(gdf_with_timepoint['timepoint'].value_counts())
```

## 面板回归

```python
def extract_panel_variables(df):
    """提取面板回归变量"""
    
    # 1. 清理和准备
    df_clean = df.dropna(subset=['rateable_value', 'lad22cd', 'timepoint'])
    
    # 2. 构建因变量
    threshold = 15000
    bandwidth = 2000
    
    df_clean['near_threshold'] = (
        (df_clean['rateable_value'] >= threshold - bandwidth) & 
        (df_clean['rateable_value'] <= threshold + bandwidth)
    ).astype(int)
    
    df_clean['log_rateable_value'] = np.log(df_clean['rateable_value'])
    
    # 3. 处理变量
    df_clean['post_revaluation'] = df_clean['timepoint'].str.contains('post').astype(int)
    df_clean['revaluation_dummy'] = (df_clean['timepoint'] == 'revaluation').astype(int)
    
    # 4. 控制变量
    df_clean['log_floor_area'] = np.log(df_clean['total_floor_area'].fillna(1) + 1)
    df_clean['occupied'] = (df_clean['occupation_state'] == 'OCCUPIED').astype(int)
    
    # 5. 固定效应变量
    area_dummies = pd.get_dummies(df_clean['lad22cd'], prefix='area')
    time_dummies = pd.get_dummies(df_clean['timepoint'], prefix='time')
    category_dummies = pd.get_dummies(df_clean['category_group'], prefix='category')
    
    # 6. 合并所有变量
    panel_df = pd.concat([
        df_clean[['lad22cd', 'timepoint', 'series', 'epoch',
                 'near_threshold', 'log_rateable_value', 
                 'post_revaluation', 'revaluation_dummy',
                 'log_floor_area', 'occupied', 'rateable_value']],
        area_dummies,
        time_dummies, 
        category_dummies
    ], axis=1)
    
    return panel_df


panel_data = extract_panel_variables(business17)
```

```python
def aggregate_by_lsoa_timepoint(panel_data):
    """按LSOA和timepoint聚合数据"""
    
    # 按LSOA-时间聚合
    lsoa_panel = panel_data.groupby(['lsoa21cd', 'timepoint']).agg({
        # 基础指标
        'near_threshold': 'mean',           # bunching比例
        'rateable_value': ['mean', 'count'], # 平均值和企业数
        'log_floor_area': 'mean',           # 平均面积
        'occupied': 'mean',                 # 占用率
        'post_revaluation': 'first',        # 时间虚拟变量
        'revaluation_dummy': 'first',       # 重估虚拟变量
        
        # 行业分类占比
        'category_group': lambda x: x.value_counts(normalize=True).to_dict()
    }).reset_index()
    
    return lsoa_panel
```

```python
def fix_boolean_variables(panel_data):
    """将布尔型变量转换为数值型"""
    
    panel_fixed = panel_data.copy()
    
    # 找到所有布尔型列
    bool_cols = []
    for col in panel_fixed.columns:
        if panel_fixed[col].dtype == 'bool':
            bool_cols.append(col)
    
    print(f"需要转换的布尔型列: {bool_cols}")
    
    # 转换为0/1
    for col in bool_cols:
        panel_fixed[col] = panel_fixed[col].astype(int)
        print(f"转换 {col}: {panel_fixed[col].unique()}")
    
    return panel_fixed

# 修复数据
panel_data_fixed = fix_boolean_variables(panel_data)

# 验证转换结果
print("\n转换后的数据类型:")
bool_related_cols = ['post_revaluation', 'category_INDUSTRIAL', 'category_LEISURE', 
                    'category_OFFICE', 'category_OTHER', 'category_RETAIL']

for col in bool_related_cols:
    if col in panel_data_fixed.columns:
        print(f"{col}: {panel_data_fixed[col].dtype}, 唯一值: {panel_data_fixed[col].unique()}")
```

```python
def create_lsoa_panel_final(panel_data):
    """使用正确列名创建LSOA面板"""
    
    # 1. 基础聚合（使用lad22cd作为地理单位）
    lsoa_agg = panel_data.groupby(['lad22cd', 'timepoint']).agg({
        'near_threshold': 'mean',
        'rateable_value': ['mean', 'std', 'count'],
        'log_floor_area': 'mean',
        'occupied': 'mean',
        'post_revaluation': 'first',
        'revaluation_dummy': 'first',
        'series': 'first',
        'epoch': 'first',
        
        # 行业分类占比（您已经有虚拟变量了！）
        'category_INDUSTRIAL': 'mean',
        'category_LEISURE': 'mean', 
        'category_OFFICE': 'mean',
        'category_OTHER': 'mean',
        'category_RETAIL': 'mean'
    }).reset_index()
    
    # 2. 展平多级列名
    new_columns = []
    for col in lsoa_agg.columns:
        if isinstance(col, tuple):
            if col[1] == '':
                new_columns.append(col[0])
            else:
                new_columns.append(f"{col[0]}_{col[1]}")
        else:
            new_columns.append(col)
    
    lsoa_agg.columns = new_columns
    
    # 3. 创建时间顺序
    timepoint_order = ['pre_12m', 'pre_6m', 'revaluation', 'post_6m', 'post_12m', 'post_24m']
    lsoa_agg['timepoint_num'] = lsoa_agg['timepoint'].map({tp: i for i, tp in enumerate(timepoint_order)})
    
    # 4. 排序
    lsoa_panel = lsoa_agg.sort_values(['lad22cd', 'timepoint_num'])
    
    return lsoa_panel

# 创建面板
lsoa_panel = create_lsoa_panel_final(panel_data)
print(f"面板形状: {lsoa_panel.shape}")
print(f"LAD数量: {lsoa_panel['lad22cd'].nunique()}")
print(f"时间点数量: {lsoa_panel['timepoint'].nunique()}")
print(f"\n列名: {lsoa_panel.columns.tolist()}")
```

```python
def fix_all_data_types(lsoa_panel):
    """彻底修复所有数据类型问题"""
    
    panel_clean = lsoa_panel.copy()
    
    print("修复前数据类型问题诊断:")
    print(panel_clean.dtypes.value_counts())
    
    # 1. 找出所有object类型的列
    object_cols = panel_clean.select_dtypes(include=['object']).columns.tolist()
    print(f"Object类型列: {object_cols}")
    
    # 2. 检查每个object列的内容
    for col in object_cols:
        unique_vals = panel_clean[col].unique()
        print(f"{col}的唯一值: {unique_vals[:5]}")  # 只显示前5个
    
    # 3. 强制转换数值相关的列
    numeric_keywords = ['threshold', 'revaluation', 'category', 'floor', 'occupied', 'rateable']
    
    for col in panel_clean.columns:
        if any(keyword in col.lower() for keyword in numeric_keywords):
            try:
                # 先尝试直接转换
                panel_clean[col] = pd.to_numeric(panel_clean[col], errors='coerce')
                print(f"✅ 成功转换 {col} 为数值型")
            except:
                # 如果失败，检查是否是布尔型
                if panel_clean[col].dtype == 'bool':
                    panel_clean[col] = panel_clean[col].astype(int)
                    print(f"✅ 成功转换 {col} 从bool到int")
                else:
                    print(f"❌ 无法转换 {col}")
    
    # 4. 删除仍然是object的数值列
    remaining_objects = panel_clean.select_dtypes(include=['object']).columns.tolist()
    exclude_cols = ['lad22cd', 'timepoint']  # 保留这些文本列
    
    problem_cols = [col for col in remaining_objects if col not in exclude_cols]
    if problem_cols:
        print(f"删除problematic列: {problem_cols}")
        panel_clean = panel_clean.drop(columns=problem_cols)
    
    # 5. 最终检查
    print("\n修复后数据类型:")
    print(panel_clean.dtypes.value_counts())
    
    return panel_clean

# 修复数据
lsoa_panel_clean = fix_all_data_types(lsoa_panel)
```

```python
def add_policy_variable(lsoa_panel):
    """重新添加政策变量"""
    
    # 基于timepoint创建post_revaluation
    lsoa_panel['post_revaluation'] = lsoa_panel['timepoint'].str.contains('post').astype(int)
    
    # 或更细致的处理
    timepoint_mapping = {
        'pre_12m': 0, 'pre_6m': 0, 'revaluation': 0,
        'post_6m': 1, 'post_12m': 1, 'post_24m': 1
    }
    lsoa_panel['post_revaluation'] = lsoa_panel['timepoint'].map(timepoint_mapping)
    
    return lsoa_panel

lsoa_panel=add_policy_variable(lsoa_panel)
```

```python
def full_panel_regression_fixed(lsoa_panel):
    """修复版完整面板回归：解决数据类型问题"""
    
    print("🔧 开始修复数据类型...")
    
    # 1. 数据类型诊断
    print("原始数据类型分布:")
    print(lsoa_panel.dtypes.value_counts())
    
    # 2. 创建干净的副本
    panel_clean = lsoa_panel.copy()
    
    # 3. 修复因变量
    bunching_cols = [col for col in panel_clean.columns if 'near_threshold' in col]
    if not bunching_cols:
        print("❌ 找不到near_threshold列")
        return None
    
    bunching_col = bunching_cols[0]
    print(f"因变量: {bunching_col}")
    
    # 强制转换因变量为数值型
    panel_clean[bunching_col] = pd.to_numeric(panel_clean[bunching_col], errors='coerce')
    y = panel_clean[bunching_col]
    print(f"因变量类型: {y.dtype}")
    
    # 4. 修复控制变量
    control_vars = []
    
    # 政策变量
    if 'post_revaluation' in panel_clean.columns:
        panel_clean['post_revaluation'] = pd.to_numeric(panel_clean['post_revaluation'], errors='coerce')
        control_vars.append('post_revaluation')
        print(f"✅ post_revaluation: {panel_clean['post_revaluation'].dtype}")
    
    # 行业变量 - 强制转换
    industry_vars = ['category_INDUSTRIAL', 'category_LEISURE', 'category_OFFICE', 'category_RETAIL']
    for var in industry_vars:
        if var in panel_clean.columns:
            # 先检查是否是布尔型
            if panel_clean[var].dtype == 'bool':
                panel_clean[var] = panel_clean[var].astype(int)
            else:
                panel_clean[var] = pd.to_numeric(panel_clean[var], errors='coerce')
            control_vars.append(var)
            print(f"✅ {var}: {panel_clean[var].dtype}")
    
    # 其他控制变量 - 安全转换
    other_control_keywords = ['floor_area', 'occupied', 'rateable_value_mean']
    for col in panel_clean.columns:
        if any(keyword in col for keyword in other_control_keywords):
            try:
                panel_clean[col] = pd.to_numeric(panel_clean[col], errors='coerce')
                control_vars.append(col)
                print(f"✅ {col}: {panel_clean[col].dtype}")
            except:
                print(f"❌ 无法转换 {col}")
    
    print(f"\n选定的控制变量: {control_vars}")
    
    # 5. 检查并删除缺失值
    analysis_vars = [bunching_col] + control_vars
    before_dropna = len(panel_clean)
    panel_clean = panel_clean.dropna(subset=analysis_vars)
    after_dropna = len(panel_clean)
    print(f"删除缺失值: {before_dropna} → {after_dropna} 行")
    
    if len(panel_clean) == 0:
        print("❌ 删除缺失值后没有数据了")
        return None
    
    # 6. 构建自变量矩阵
    X = panel_clean[control_vars]
    print(f"X矩阵形状: {X.shape}")
    print(f"X数据类型: {X.dtypes.value_counts()}")
    
    # 7. 创建固定效应 - 确保数值型
    print("\n创建固定效应...")
    
    # LAD固定效应
    lad_fe = pd.get_dummies(panel_clean['lad22cd'], prefix='lad', dtype=int)
    print(f"LAD固定效应: {lad_fe.shape[1]}个")
    
    # 时间固定效应  
    time_fe = pd.get_dummies(panel_clean['timepoint'], prefix='time', dtype=int)
    print(f"时间固定效应: {time_fe.shape[1]}个")
    
    # 8. 合并所有变量
    X_full = pd.concat([
        X,
        lad_fe.iloc[:, :-1],    # 排除一个LAD避免完全多重共线性
        time_fe.iloc[:, :-1]    # 排除一个时间
    ], axis=1)
    
    # 9. 最终数据类型检查
    print(f"\n合并后X_full形状: {X_full.shape}")
    
    # 确保所有列都是数值型
    non_numeric = X_full.select_dtypes(exclude=[np.number]).columns
    if len(non_numeric) > 0:
        print(f"❌ 仍有非数值列: {non_numeric.tolist()}")
        # 强制转换或删除
        for col in non_numeric:
            try:
                X_full[col] = pd.to_numeric(X_full[col], errors='coerce')
            except:
                X_full = X_full.drop(columns=[col])
                print(f"删除列: {col}")
    
    # 10. 添加常数项
    X_full = sm.add_constant(X_full)
    
    # 11. 最终验证
    print(f"\n最终数据验证:")
    print(f"y形状: {y.shape}, 类型: {y.dtype}")
    print(f"X_full形状: {X_full.shape}")
    print(f"y有缺失值: {y.isnull().sum()}")
    print(f"X_full有缺失值: {X_full.isnull().sum().sum()}")
    
    # 删除任何剩余的缺失值
    final_data = pd.concat([y, X_full], axis=1).dropna()
    y_final = final_data.iloc[:, 0]
    X_final = final_data.iloc[:, 1:]
    
    print(f"最终样本量: {len(y_final)}")
    
    # 12. 运行回归
    try:
        model = sm.OLS(y_final, X_final).fit()
        print("\n=== 完整面板回归结果 ===")
        print(model.summary())
        return model
    except Exception as e:
        print(f"❌ 回归失败: {e}")
        print("尝试诊断...")
        
        # 诊断信息
        print(f"y_final样本: {y_final.head()}")
        print(f"X_final数据类型: {X_final.dtypes.value_counts()}")
        
        return None

# 运行修复版回归
full_model = full_panel_regression_fixed(lsoa_panel)
```
